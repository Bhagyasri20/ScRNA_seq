```python
import cobra
```

```python
model = cobra.io.read_sbml_model("Helicobacter_pylori.xml")
```

```python
# want to know what the model contains)
print("Model Id : ", model.id)
print("Number of Reactions : ", len(model.reactions))
print("Number of Metabolites : ", len(model.metabolites))
print("Number of Genes : ", len(model.genes))
print("Number of Compartments : " , len(model.compartments))
```

```python
print(model.objective)
```

```python
for comp_id, comp_name in model.compartments.items():
    print(comp_id, ":" , comp_name)
```

```python
# checking for reaction bounds
invalid_bounds = []

for rxn in model.reactions:
    if rxn.lower_bound > rxn.upper_bound:
        invalid_bounds.append(rxn.id)

print("Invalid bounds : ", len(invalid_bounds))

if invalid_bounds:
    print(invalid_bounds)
else:
    print("All reaction bounds are valid.")
```

```python
from collections import Counter

# Reaction IDs
reaction_ids = [rxn.id for rxn in model.reactions]
reaction_counts = Counter(reaction_ids)

duplicate_reactions = {
    rid: count for rid, count in reaction_counts.items()
    if count > 1
}

# Metabolite IDs
metabolite_ids = [met.id for met in model.metabolites]
metabolite_counts = Counter(metabolite_ids)

duplicate_metabolites = {
    mid: count for mid, count in metabolite_counts.items()
    if count > 1
}

# Gene IDs
gene_ids = [gene.id for gene in model.genes]
gene_counts = Counter(gene_ids)

duplicate_genes = {
    gid: count for gid, count in gene_counts.items()
    if count > 1
}

print("Duplicate reaction IDs:")
print(duplicate_reactions)

print("\nDuplicate metabolite IDs:")
print(duplicate_metabolites)

print("\nDuplicate gene IDs:")
print(duplicate_genes)
```
