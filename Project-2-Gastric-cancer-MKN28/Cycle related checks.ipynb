```python
# to check if model has internal thermodynamically suspicious cycles
import cobra
model = cobra.io.read_sbml_model("Helicobacter_pylori.xml")
```

```python
from cobra.flux_analysis import loopless_solution

loopless_sol = loopless_solution(model)

print("Loopless FBA status:", loopless_sol.status)
print("Loopless biomass flux:", loopless_sol.objective_value)
```

```python
# normal and loopless FBA give essentially same biomass flux suggesting optimal biomass sol doesnot
# depend on detectable internal flux loops
```
